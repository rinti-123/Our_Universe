```python
import numpy as np
import matplotlib.pyplot as plt

# Create a grid of points
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Define the vector field F(x,y) = <x^2, y^2>
U = -Y
V = X

# Plot the vector field
plt.figure(figsize=(8, 8))
plt.quiver(X, Y, U, V, color='blue', angles='xy', scale_units='xy', scale=8)

# Plot the line y = x
t = np.linspace(-2, 2, 200)
plt.plot(t, -
t, 'r-', linewidth=2, label='y = x')

# Labels and formatting
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()

plt.show()
```

```python
%matplotlib qt
import numpy as np
import matplotlib.pyplot as plt

# Create a 3D grid
x = np.linspace(-2, 2, 6)
y = np.linspace(-2, 2, 6)
z = np.linspace(-2, 2, 6)

X, Y, Z = np.meshgrid(x, y, z)

# Define the vector field F(x,y,z) = <x², y², z²>
U = X**2
V = Y**2
W = Z**2

# Create 3D figure
fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the vector field
ax.quiver(
    X, Y, Z,
    U, V, W,
    length=0.15,        # Arrow length
    normalize=True,     # Normalize arrow lengths
    color='blue'
)

# Labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title(r'3D Vector Field $\mathbf{F}(x,y,z)=\langle x^2,y^2,z^2\rangle$')

# Set axis limits
ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.set_zlim(-2, 2)

plt.show()
```
